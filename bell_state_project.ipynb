```python
# Import QuantumCircuit: this is our "blank canvas" where we build quantum programs.
from qiskit import QuantumCircuit

# Import AerSimulator: this is our fake (simulated) quantum computer.
from qiskit_aer import AerSimulator

# Import plot_histogram: a helper that draws our results as a bar chart.
from qiskit.visualization import plot_histogram

# Import pyplot: used to display the chart on the screen.
import matplotlib.pyplot as plt
```

```python
# Create a circuit with 2 qubits and 2 classical bits.
# A qubit is the quantum version of a bit; unlike a normal bit (0 or 1), it can be in a mix of both.
# Classical bits are normal storage boxes where we save the final measured answers (0 or 1).
qc = QuantumCircuit(2, 2)

# Apply a Hadamard (H) gate to qubit 0.
# The Hadamard gate puts a qubit into "superposition": a 50% chance of 0 and a 50% chance of 1 when measured.
qc.h(0)

# Apply a CNOT gate with qubit 0 as the "control" and qubit 1 as the "target".
# CNOT flips qubit 1 only if qubit 0 is 1. Because qubit 0 is in superposition, this links the two qubits together.
# This linking is called ENTANGLEMENT, and the result is a Bell State.
qc.cx(0, 1)

# Measure both qubits and store the results in the classical bits.
# Measuring forces each qubit to "choose" 0 or 1. Qubit 0 goes to bit 0, qubit 1 goes to bit 1.
qc.measure([0, 1], [0, 1])
```

```python
# Draw the circuit as text so we can see the gates in order (left to right = time).
# "text" mode works everywhere with no extra setup.
print(qc.draw("text"))
```

```python
# Create our simulated quantum computer.
simulator = AerSimulator()

# Run the circuit 1000 times. Each run is called a "shot".
# We need many shots because quantum results are random, so we look at the overall pattern.
job = simulator.run(qc, shots=1000)

# Wait for the simulator to finish and collect everything it produced.
result = job.result()

# Get the counts: how many times each answer (like "00" or "11") appeared.
counts = result.get_counts()

# Print the counts so we can read the raw numbers.
print("Results:", counts)
```

```python
# Draw a bar chart of the counts.
plot_histogram(counts)

# Display the chart on screen.
plt.show()
```
